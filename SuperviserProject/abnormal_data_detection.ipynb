```python
def record_duplication(year, month):
    root_path = "D:/yuanjun/SuperviserProject\data\original"
    df = pd.read_csv(os.path.join(root_path, "original_%s%s.csv" % (year, month)))
    origin_record_num = df.shape[0]
    df.drop_duplicates(["basy_sfzh_md5", "basy_rysj", "basy_cysj", "basy_zfy"], keep='first', inplace=False)
    duplicates_num = df.shape[0]
    print("%s%s重复数据数量：" % (year, month), origin_record_num-duplicates_num)
    return origin_record_num, origin_record_num-duplicates_num
result_list = list()
for month in ["01", "02", "03", "04", "05", "06", "07", "08", "09", "10", "11", "12"]:
    month_record, duplication_num = record_duplication("2023", month)
    result_list.append([month_record, duplication_num])
```

```python
def draw_month_count():
    record_count_dict = {
        "month": [],
        "count": []
    }
    for i in range(len(result_list)):
        record_count_dict["month"].append(i)
        record_count_dict["count"].append(result_list[i][0])
    df = pd.DataFrame(record_count_dict)
    df.plot(x="month", y="count", xlabel="month", ylabel="count")
draw_month_count()
```

```python
def clean_birthday(birthday, year=2023):
    """
    生日清洗，转化为年龄，大于100岁默认100岁,其余不合法的默认为空
    :param birthday: 出生日期（value_feature_353）
    :param year: 当前年份
    :return: 当前年份的年龄
    """
    birthday = str(birthday)
    if len(birthday) > 4:
        age = year - int(birthday[:4])
        return age
    else:
        return np.nan

def month_age_distribution():
    month_age_dict = dict()
    miss_value_dict = dict()
    root_path = "D:\yuanjun\SuperviserProject\data\original"
    for i in os.listdir(root_path):
        if i == "hospital_info.csv" or i == 'hospital_info_V1.csv':
            continue
        
        month_age_dict[i[9: -4]] = list()
        df = pd.read_csv(os.path.join(root_path, i), usecols=["basy_csrq"])
        df["AGE"] = df["basy_csrq"].map(lambda a: clean_birthday(a))
        df["count"] = 1
        age_group = df.groupby("AGE").agg({'count': 'count'}).reset_index()
#         print(age_group.values)
        age_count_dict = dict()
        for age_count in age_group.values:
            age_count_dict[age_count[0]] = age_count[1]
#         print(age_count_dict)
        for age in range(130):
            month_age_dict[i[9: -4]].append(age_count_dict.get(age, 0))
        num = df["AGE"].isna().sum()
        if num > 0:
#             print(i)
            df["ab_normal"] = df["AGE"].isna()
            ab_normal_df = df[df["ab_normal"]]
            print("ab_normal_df", ab_normal_df)
        print("MiSS VALUE COUNT:", num)
#         print(month_age_dict)
        miss_value_dict[i] = num
    return month_age_dict, miss_value_dict

month_age_dict, miss_value_dict = month_age_distribution()  
```

```python
def month_sjzy_distribution():
    month_sjzy_dict = dict()
    miss_value_dict = dict()
    root_path = "D:\yuanjun\SuperviserProject\data\original"
    for i in tqdm(os.listdir(root_path)):
        if i == "hospital_info.csv" or i == 'hospital_info_V1.csv':
            continue
        month_sjzy_dict[i[9: -4]] = list()
        df = pd.read_csv(os.path.join(root_path, i), usecols=["basy_sjzy"])
        num = df["basy_sjzy"].isna().sum()
        if num > 0:
            print("MiSS VALUE COUNT:", i, num)
        df["count"] = 1
        age_group = df.groupby("basy_sjzy").agg({'count': 'count'}).reset_index()
        age_count_dict = dict()
        for age_count in age_group.values:
            age_count_dict[age_count[0]] = age_count[1]

        for age in range(30):
            month_sjzy_dict[i[9: -4]].append(age_count_dict.get(age, 0))
        
    return month_sjzy_dict

month_sjzy_dict = month_sjzy_distribution()
```

```python
bm_list = ["basy_zyzd_jbbm1", "basy_zyzd_jbbm2", "basy_zyzd_jbbm3", "basy_zyzd_jbbm4", "basy_zyzd_jbbm5"
                                                             , "basy_zyzd_jbbm6", "basy_zyzd_jbbm7", "basy_zyzd_jbbm8", "basy_zyzd_jbbm9", "basy_zyzd_jbbm10"
                                                             ,"basy_zyzd_jbbm11", "basy_zyzd_jbbm12", "basy_zyzd_jbbm13", "basy_zyzd_jbbm14", "basy_zyzd_jbbm15"]
def cal_jbbm_count(a):
    count = 0
    for bm in bm_list:
        tmp_v = a[bm]
        if isinstance(tmp_v, str):
            tmp_v = ''.join(filter(str.isalnum, tmp_v))
            if len(tmp_v) > 2:
                count += 1
    return count
            
def qita_zd_num_dis():
    month_zd_num_dict = dict()
    root_path = "D:\yuanjun\SuperviserProject\data\original"
    for i in tqdm(os.listdir(root_path)):
        if i == "hospital_info.csv" or i == 'hospital_info_V1.csv':
            continue
        month_zd_num_dict[i[9: -4]] = list()
        df = pd.read_csv(os.path.join(root_path, i), usecols=bm_list)
        df["jb_total_count"] = df.apply(cal_jbbm_count, axis=1)
        df["count"] = 1
        zd_num_group = df.groupby("jb_total_count").agg({'count': 'count'}).reset_index()
        zd_count_dict = dict()
        for zd_num in zd_num_group.values:
            zd_count_dict[zd_num[0]] = zd_num[1]

        for age in range(16):
            month_zd_num_dict[i[9: -4]].append(zd_count_dict.get(age, 0))
    return month_zd_num_dict
month_zd_num_dict = qita_zd_num_dis()

        
```
